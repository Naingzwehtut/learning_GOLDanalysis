# 02 — Data Cleaning

Loads the raw CSV from `data/raw/`, cleans it using `scripts/clean_data.py`, and saves an analysis-ready version to `data/processed/`.

```python
import sys
sys.path.append('../scripts')
import pandas as pd

from clean_data import clean_gold_data, save_processed_data
```

```python
df_raw = pd.read_csv('../data/raw/xauusd_raw.csv')
df_raw.shape
```

## Run the cleaning pipeline

This removes duplicates, handles missing values, converts date types, sorts ascending, and validates required OHLCV columns.

```python
df_clean = clean_gold_data(df_raw)
df_clean.head()
```

## Sanity checks

```python
print('Rows before:', len(df_raw), '| Rows after:', len(df_clean))
print('Any nulls remaining?', df_clean.isna().any().any())
print('Date range:', df_clean['Date'].min(), '->', df_clean['Date'].max())
```

## Save cleaned data

```python
save_processed_data(df_clean, filename='xauusd_clean.csv')
```
